```python
import pandas as pd
```

```python
certificates_earned = pd.DataFrame({'Certificates': [8, 2, 5, 6], 'Time (in months)': [16, 5, 9, 12]});
certificates_earned.index = ['Tom', 'Kris', 'Ahmad', 'Beau'];
print(certificates_earned)
```

```python
print(certificates_earned[certificates_earned['Certificates']>5])
```

```python
certificates_earned['Certificates per month'] = round(certificates_earned['Certificates'] / certificates_earned['Time (in months)'], 2)
print(certificates_earned)
```
